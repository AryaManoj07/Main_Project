Outlier Detection Using IQR
</h2>



```python
def find_outliers(column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    
    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]
    
    print("Column:", column)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower limit:", lower_limit)
    print("Upper limit:", upper_limit)
    print("Number of outliers:", len(outliers))
    
    return outliers
```

## Classification of EDA

### Univariate Analysis
Analysis of one variable at a time.

Examples:
- Distribution of movie ratings
- Distribution of movie budgets
- Distribution of revenue

### Bivariate Analysis
Analysis of the relationship between two variables.

Examples:
- Budget vs Revenue
- Budget vs Profit
- Popularity vs Revenue
- Revenue by Genre
- Revenue Trend Over Time

### Multivariate Analysis
Analysis involving three or more variables.

Example:
- Correlation Heatmap



Here, univariate analysis to understand individual variables, bivariate analysis to compare relationships between two variables, and multivariate analysis to study relationships among several numerical variables.

```python
interactive_df = df[
    (df['budget'] > 0) &
    (df['revenue'] > 0)
].copy()

fig = px.scatter(
    interactive_df.sample(
        min(3000, len(interactive_df)),
        random_state=42
    ),
    x='budget',
    y='revenue',
    hover_name='title',
    hover_data=[
        'vote_average',
        'popularity',
        'runtime'
    ],
    title='Interactive Budget vs Revenue',
    labels={
        'budget': 'Budget',
        'revenue': 'Revenue'
    }
)

fig.update_layout(
    width=1000,
    height=600
)

fig.show()
```

However, the relationship is not perfect. Some movies with relatively high budgets do not generate equally high revenue, while some movies achieve strong revenue with comparatively lower budgets.

This shows that budget is an important factor associated with commercial performance, but it is not the only factor determining movie success.

The interactive Plotly chart makes this relationship easier to explore because individual movies can be identified by hovering over the points.

# Interactive Genre Revenue

```python
genre_plot_df = df.dropna(subset=['genres']).copy()

genre_plot_df['genre'] = genre_plot_df['genres'].str.split(',')

genre_plot_df = genre_plot_df.explode('genre')

genre_plot_df['genre'] = genre_plot_df['genre'].str.strip()

genre_summary = (
    genre_plot_df
    .groupby('genre')
    .agg(
        average_revenue=('revenue', 'mean'),
        movie_count=('title', 'count')
    )
    .reset_index()
)

genre_summary = genre_summary[
    genre_summary['movie_count'] >= 100
]

genre_summary = genre_summary.sort_values(
    'average_revenue',
    ascending=False
).head(15)

fig = px.bar(
    genre_summary,
    x='genre',
    y='average_revenue',
    hover_data=['movie_count'],
    title='Interactive Average Revenue by Genre',
    labels={
        'genre': 'Genre',
        'average_revenue': 'Average Revenue'
    }
)

fig.update_layout(
    width=1000,
    height=600
)

fig.show()
```



The interactive genre analysis shows that average revenue differs considerably across movie genres.

Adventure has the highest average revenue among the genres included in the analysis, followed by Fantasy and Science Fiction.

This indicates that some genres have historically generated higher average revenue within this dataset. However, this does not mean that genre alone causes higher revenue, because movie performance can also depend on factors such as budget, popularity, cast, audience interest and other characteristics.

The interactive chart also displays the number of movies represented by each genre, which provides useful context when comparing the average revenue of different genres.

# Interactive Revenue Trend

```python
trend_df = df.copy()

trend_df['release_date'] = pd.to_datetime(
    trend_df['release_date'],
    errors='coerce'
)

trend_df['release_year'] = trend_df['release_date'].dt.year

trend_df = trend_df[
    (trend_df['release_year'] >= 1950) &
    (trend_df['release_year'] <= 2025) &
    (trend_df['revenue'] > 0)
]

year_summary = (
    trend_df
    .groupby('release_year')
    .agg(
        average_revenue=('revenue', 'mean'),
        movie_count=('title', 'count')
    )
    .reset_index()
)

fig = px.line(
    year_summary,
    x='release_year',
    y='average_revenue',
    markers=True,
    hover_data=['movie_count'],
    title='Interactive Movie Revenue Trend Over Time',
    labels={
        'release_year': 'Release Year',
        'average_revenue': 'Average Revenue'
    }
)

fig.update_layout(
    width=1000,
    height=600
)

fig.show()
```



The interactive revenue trend shows that average movie revenue changes substantially across different release years rather than increasing continuously over time.

The trend contains periods of higher and lower average revenue, with a noticeable decline around 2020 followed by a recovery in the following years.

This indicates that movie revenue can be affected by changes in the movie industry and external circumstances, rather than simply increasing with time.

The Plotly version makes the trend easier to explore because the average revenue and number of movies can be viewed for individual years by hovering over the data points.
